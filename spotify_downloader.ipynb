```python
import subprocess, sys, os, time
from IPython.display import clear_output

playlist_url = "https://open.spotify.com/playlist/343BDswtiUAtBPuKSs7l92?si=mLRnEqVSRNKPQNcDWooMYw"
output_folder = "downloaded_playlist"   # folder in current directory
log_file = "sunnify_log.txt"

# Clean old log
if os.path.exists(log_file):
    os.remove(log_file)

# Correct command: --quality 128 and explicit output folder
cmd = [
    sys.executable, "Spotify_Downloader.py", "download",
    playlist_url,
    "--format", "mp3",
    "--quality", "128",
    "--out", output_folder
]

# Start download in background, redirect output to log
with open(log_file, "w") as f:
    process = subprocess.Popen(cmd, stdout=f, stderr=subprocess.STDOUT, text=True)

print(f"Download started (PID {process.pid}). Output folder: {output_folder}")

try:
    while process.poll() is None:
        time.sleep(3)
        with open(log_file, "r") as f:
            lines = f.readlines()
        last_lines = lines[-10:] if len(lines) > 10 else lines

        clear_output(wait=True)
        print(f"Downloading... (PID {process.pid}) -> {output_folder}\n")
        print("".join(last_lines))
        print(f"\nStill running: {process.poll() is None}")

    clear_output(wait=True)
    print("✅ Download completed!\n")
    with open(log_file, "r") as f:
        lines = f.readlines()
    print("".join(lines[-15:]))
    print(f"\nReturn code: {process.returncode}")

except KeyboardInterrupt:
    process.terminate()
    clear_output(wait=True)
    print("Download interrupted. You can re-run this cell later to resume.")
```

```python
import os, shutil

output_folder = "downloaded_playlist"

if not os.path.isdir(output_folder):
    print("Output folder not found. Did the download complete successfully?")
else:
    zip_name = output_folder + ".zip"
    shutil.make_archive(output_folder, 'zip', output_folder)
    print("Created zip file:", zip_name)
```
